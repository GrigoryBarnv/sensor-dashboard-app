```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV laden
df = pd.read_csv("data/Example_Data.csv", sep=';')
df.columns = df.columns.str.strip()
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S")

# Alle verfügbaren Sensoren anzeigen
sensor_columns = [col for col in df.columns if col.endswith("_RS")]
sensor_ids = [col.replace("_RS", "") for col in sensor_columns]

print("🔎 Verfügbare Sensoren:")
print(", ".join(sensor_ids))  # unter dem Plot als Liste

# Sensor auswählen (manuell eingeben)
sensor_id = input("Gib den Sensornamen ein (z. B. MQ2, MQ135): ").strip()

# Plot anzeigen
column = f"{sensor_id}_RS"
if column in df.columns:
    plt.figure(figsize=(10, 4))
    plt.plot(df["Time"], df[column], label=column, color='blue')
    plt.title(f"{sensor_id} Sensorwert über Zeit")
    plt.xlabel("Zeit")
    plt.ylabel("Widerstand (Ohm)")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print(f"❌ Spalte '{column}' nicht gefunden!")
```
